```python

#Eigenvalues and eigenvectors
#linalg.eig
a = np.array([[ 0, 1],
             [ 2, 3]])
eigenvalues, eigenvectors = np.linalg.eig(a)
print("eigenvalues are: ", eigenvalues)
print("eigenvectors are: ", eigenvectors)
```

```python

#The np.linalg.solve() solves a linear matrix equation, or system of linear scalar equations.
a = np.array([[2, 2], [4, 2]])
b = np.array([6,-1])
solution = np.linalg.solve(a,b)
print(solution)
```

```python
import numpy as np
coeff=np.array([2,1,1,0,2])
f=np.poly1d(coeff)
print(f)
print("====================")
print(f(1000))
print("====================")
print(f.r)  #root
```

```python
import numpy as np
coeff=np.array([1,2,1])
f=np.poly1d(coeff)
print(f)
print("====================")
f1=np.polyder(f,1)
print(f1)
print(f1(0))
print("====================")
f2=np.polyder(f,2)
print(f2)
print("====================")
```

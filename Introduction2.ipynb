```python
# mean of your score

n=int(input("n="))
s=0
sv=0
for i in range(n):
    v=float(input("v="))  # unit
    m=float(input("x="))  # score
    x=m*v
    s=s+x
    sv=sv+v
mean=s/sv
print("mean=",mean)
```

```python
x={"name":["reza","elham","hamid"],
   "age":[32,25,28],
   "city":["esfahan","tehran","karaj"]}


S1=x["name"]
S2=x["age"]
S3=x["city"]

n=S1.index("elham")
S1.insert(n+1,"sepideh")
S2.insert(n+1,21)
S3.insert(n+1,"hamedan")

x["name"]=S1
x["age"]=S2
x["city"]=S3

print(x)
```

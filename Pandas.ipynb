```python
#DataFrame (two-dimensional data)
#pd.DataFrame() creates a table-like structure (rows & columns).
#converting a dictionary to dataframe
#Keys of the dictionary (Name, Age, Score) become column names.
#Values (lists) become the data inside each column.

data = {
    "Name": ["Ali", "Sara", "Nima"],
    "Age": [23, 19, 30],
    "Score": [85, 92, 78]
}

df = pd.DataFrame(data)
print(df)
print("---------------")
print(df.loc[2]) # the third row
print("---------------")
print(df.loc[[0,1]]) # the first and second rows
print("---------------")
print(df.loc[1]["Score"]) #the second row and the second column
```

```python
print(df.sort_values("Age"))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data1={"date":[5,10,15,20],"temp":[23,26,27,28]}
y1=pd.DataFrame(data1)
data2={"date":[1,10,20,29],"temp":[26,27,29,30]}
y2=pd.DataFrame(data2)
print(y1)
print("__________________")
print(y2)
print("__________________")
print(pd.merge(y1,y2,on="date")) # merge them based on data column
print(pd.merge(y1,y2,on="temp"))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data1={"date":[5,5,5,5,10,15,15,20],
       "temp":[23,25,27,25,26,27,30,28],
       "yy":[10,50,70,60,80,90,83,22]}
y1=pd.DataFrame(data1)

print(y1)
print("_________________")

print(y1.groupby("date").sum()) #مقدار ترکراری وجود دارد با هم جمع می شوند.  ردیف هایی که در ستون date
print("_________________")
print(y1.groupby("date").mean())
print("_________________")
print(y1.groupby("date").min())
print("_________________")
print(y1.groupby("date").max())
print("_________________")
print(y1.sort_values("temp"))
print("_________________")
print(y1.sort_values(by=["temp"]))
```

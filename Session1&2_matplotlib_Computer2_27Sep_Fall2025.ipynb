```python

# Contour plot
#---------------------------------------------------------
#Example #1: Plotting of Contour using contour() which only plots contour lines.
#---------------------------------------------------------

# Implementation of matplotlib function
import matplotlib.pyplot as plt
import numpy as np

feature_x = np.arange(0, 50, 2)
feature_y = np.arange(0, 50, 3)

# Creating 2-D grid of features
[X, Y] = np.meshgrid(feature_x, feature_y)

fig, ax = plt.subplots(1, 1)

Z = np.cos(X / 2) + np.sin(Y / 4)

# plots contour lines
ax.contour(X, Y, Z)

ax.set_title('Contour Plot')
ax.set_xlabel('feature_x')
ax.set_ylabel('feature_y')

plt.show()
```

```python
x=np.arange(-3,3,0.05)
y=np.arange(-2,2,0.05)
X,Y=np.meshgrid(x,y)
Z=np.exp(-X**4-Y**2)
fig,ax=plt.subplots()
cs=ax.contour(X,Y,Z)
cs=ax.contourf(X,Y,Z)
#ax.clabel(cs,inline=1,fontsize=10)
fig.colorbar(cs)
plt.show()
```

```python
x=np.arange(-3,3,0.05)
y=np.arange(-2,2,0.05)
X,Y=np.meshgrid(x,y)
z1=np.exp(-X**4-Y**2)
z2=np.exp(-(X-1)**4-(Y-1)**2)
Z=(z1-z2)*2
fig,ax=plt.subplots()
cs=ax.contour(X,Y,Z)
#cs=ax.contour(X,Y,Z,colors="k")
ax.clabel(cs,inline=1,fontsize=10)
fig.colorbar(cs)
plt.show()
```

```python
#Example #2: Plotting of contour using contourf() which plots filled contours.
#-----------------------------------------------------------
# Implementation of matplotlib function
import matplotlib.pyplot as plt
import numpy as np

feature_x = np.linspace(-2*np.pi, 2*np.pi, 70)
feature_y = np.linspace(-1.0, 1.0, 70)

# Creating 2-D grid of features
[X, Y] = np.meshgrid(feature_x, feature_y)

fig, ax = plt.subplots()

Z =10*np.exp(-0.1*Y)*np.cos(0.8*X)

# plots filled contour plot
ax.contourf(X, Y, Z)

ax.set_title('Filled Contour Plot')
ax.set_xlabel('feature_x')
ax.set_ylabel('feature_y')

plt.show()
```
